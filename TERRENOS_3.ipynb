#**Ejemplo 2: Sistem de recomendaciòn con variable relevante**

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
import numpy as np

# Simulación de datos de terrenos optimizados
data=pd.read_csv('/content/drive/MyDrive/basedatos/BasedeDatosDefinitivaIDRELEVANTE.csv')
data

# Cargamos los datos en un DataFrame
df = pd.DataFrame(data)

# Procesamiento de características categóricas con OneHotEncoder
encoder = OneHotEncoder()
ciudad_encoded = encoder.fit_transform(df[['ciudad']]).toarray()
zonificación_encoded = encoder.fit_transform(df[['zonificación']]).toarray()

# Escalamos las características numéricas
scaler = MinMaxScaler()
numerical_features = scaler.fit_transform(df[['area', 'precio(mts2)']])

# Combinamos las características
features = np.hstack((numerical_features, ciudad_encoded, zonificación_encoded))

# Seleccionamos un terreno de referencia (id = 1)
terreno_referencia_idx = 0
terreno_referencia = features[terreno_referencia_idx]

# Calculamos la similitud del coseno
similitudes = cosine_similarity([terreno_referencia], features)[0]

# Creamos un DataFrame con los resultados ordenados por similitud
df['similitud'] = similitudes
recomendaciones = df.sort_values(by='similitud', ascending=False)

# --- Métricas de Evaluación ---
# Seleccionamos los primeros K terrenos como recomendados
K = 3
top_k = recomendaciones.iloc[:K]

# Precision@K
precision_at_k = top_k['relevante'].sum() / K

# Recall@K
total_relevantes = df['relevante'].sum()
recall_at_k = top_k['relevante'].sum() / total_relevantes if total_relevantes > 0 else 0

# F1-Score
f1_score = (2 * precision_at_k * recall_at_k) / (precision_at_k + recall_at_k) if (precision_at_k + recall_at_k) > 0 else 0

# Reporte
print("Reporte de rendimiento:")
print(f"Precision@{K}: {precision_at_k:.2f}")
print(f"Recall@{K}: {recall_at_k:.2f}")
print(f"F1-Score: {f1_score:.2f}")

# Recomendaciones ordenadas
print("\nTerrenos recomendados basados en la similitud:")
print(recomendaciones[['id', 'ciudad', 'area', 'zonificación', 'precio(mts2)', 'similitud', 'relevante']])
```

Informe de rendimiento:
Precisión@3: 0,67 significa que de las 3 principales recomendaciones, 2 fueron realmente relevantes (2/3 = 0,67).
Recordatorio@3: 0,20 indica que de todas las tierras relevantes en el conjunto de datos, solo el 20 % se incluyeron en las 3 principales recomendaciones.
Puntuación F1: 0,31 es una puntuación combinada que muestra un equilibrio moderado entre precisión y recordación.
Terrenos recomendados:
Esta tabla muestra las 3 principales tierras recomendadas (id 1, 4 y 3) con sus características y puntuaciones de similitud.
La columna relevante indica si una tierra recomendada fue realmente relevante (1 para relevante, 0 para no relevante).

#**Ejemplo 3: Sistema de recomendciòn considerando la relevancia como escala de likert**

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
import numpy as np

# Simulación de datos de terrenos con escala Likert de relevancia

data=pd.read_csv('/content/drive/MyDrive/basedatos/BasedeDatosDefinitivaIrelevancia_likert.csv')
data
# Cargamos los datos en un DataFrame
df = pd.DataFrame(data)

# Procesamiento de características categóricas con OneHotEncoder
encoder = OneHotEncoder()
ciudad_encoded = encoder.fit_transform(df[['ciudad']]).toarray()
tipo_encoded = encoder.fit_transform(df[['zonificación']]).toarray()

# Escalamos las características numéricas
scaler = MinMaxScaler()
numerical_features = scaler.fit_transform(df[['area', 'precio(mts2)']])

# Combinamos las características
features = np.hstack((numerical_features, ciudad_encoded, tipo_encoded))

# Seleccionamos un terreno de referencia (id = 1)
terreno_referencia_idx = 0
terreno_referencia = features[terreno_referencia_idx]

# Calculamos la similitud del coseno
similitudes = cosine_similarity([terreno_referencia], features)[0]

# Añadimos la similitud al DataFrame y ordenamos por relevancia
df['similitud'] = similitudes

# Aseguramos que la columna 'relevante' exista correctamente
relevancia_threshold = 4
df['relevante'] = (df['relevancia_likert'] >= relevancia_threshold).astype(int)

recomendaciones = df.sort_values(by='similitud', ascending=False)

# --- Métricas de Evaluación ---
# Definimos un umbral para considerar relevante (Likert >= 4)
#relevancia_threshold = 4

# Aseguramos que la columna 'relevante' exista correctamente
#df['relevante'] = (df['relevancia_likert'] >= relevancia_threshold).astype(int)

# Seleccionamos los primeros K terrenos como recomendados
K = 3
top_k = recomendaciones.iloc[:K]

# Precision@K
precision_at_k = top_k['relevante'].sum() / K

# Recall@K
total_relevantes = df['relevante'].sum()
recall_at_k = top_k['relevante'].sum() / total_relevantes if total_relevantes > 0 else 0

# F1-Score
f1_score = (2 * precision_at_k * recall_at_k) / (precision_at_k + recall_at_k) if (precision_at_k + recall_at_k) > 0 else 0

# Weighted Precision: ponderamos por la escala Likert
weighted_precision = top_k.apply(lambda x: x['relevancia_likert'] if x['relevante'] == 1 else 0, axis=1).sum() / (K * 5)

# Reporte
print("Reporte de rendimiento:")
print(f"Precision@{K}: {precision_at_k:.2f}")
print(f"Recall@{K}: {recall_at_k:.2f}")
print(f"F1-Score: {f1_score:.2f}")
print(f"Weighted Precision: {weighted_precision:.2f}")

# Recomendaciones ordenadas
print("\nTerrenos recomendados basados en la similitud:")
print(recomendaciones[['id', 'ciudad', 'area', 'zonificación', 'precio(mts2)', 'similitud', 'relevancia_likert', 'relevante']])
```

Precision@3: 1.00: Esto significa que las 3 recomendaciones principales (K=3) son relevantes para el usuario. En otras palabras, todas las recomendaciones proporcionadas por el sistema son acertadas.

Recall@3: 1.00: Esto significa que todos los terrenos relevantes en el conjunto de datos fueron incluidos en las 3 principales recomendaciones. El sistema no ha pasado por alto ningún terreno relevante.

F1-Score: 1.00: El F1-Score es una medida que combina la precisión y el recall. Un valor de 1.00 indica un rendimiento perfecto, lo que significa que el sistema tiene una precisión y un recall perfectos.

Weighted Precision: 0.93: Esta métrica es similar a la precisión, pero ponderada por la escala Likert, dando más peso a los terrenos con mayor relevancia. Un valor de 0.93 indica que las recomendaciones tienen una alta relevancia, aunque no perfecta, teniendo en cuenta la escala Likert.
